```python
from google import genai

# ==========================
# CẤU HÌNH
# ==========================

API_KEY = "KEY"

client = genai.Client(api_key=API_KEY)

MODEL = "gemini-2.5-flash"

SYSTEM_PROMPT = """
Bạn là nhân viên tư vấn cho thuê phòng.

Vai trò của bạn:

- Trả lời như một nhân viên thật.
- Luôn lịch sự, thân thiện.
- Không bịa thông tin.
- Nếu chưa đủ dữ liệu thì hỏi lại khách.
- Không tự ý khẳng định còn phòng.
- Không tự ý đưa ra giá nếu khách chưa cung cấp dữ liệu.
- Trả lời ngắn gọn, tự nhiên như đang chat Zalo.
"""

chat = client.chats.create(
    model=MODEL,
    config={
        "system_instruction": SYSTEM_PROMPT
    }
)

print("=" * 60)
print("Gemini Chat")
print("Nhập exit để thoát.")
print("=" * 60)

while True:

    user = input("\nKhách: ")

    if user.lower() == "exit":
        break

    response = chat.send_message(user)

    print("\nNhân viên:")
    print(response.text)
```
